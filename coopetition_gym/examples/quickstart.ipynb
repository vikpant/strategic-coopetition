# Coopetition-Gym Quickstart

This notebook demonstrates the basic API of `coopetition_gym`. We load a multi-agent environment, take random actions for one episode, and inspect the rewards.

Coopetition-Gym implements 20 multi-agent reinforcement learning environments for studying mixed-motive coopetitive dynamics, organised into four mechanism classes: interdependence (TR-1), trust (TR-2), collective action (TR-3), and reciprocity (TR-4).

## Single-agent (Gymnasium) API

The Gymnasium adapter returns observations and rewards as arrays indexed by agent.

```python
import coopetition_gym
import numpy as np

env = coopetition_gym.make('TrustDilemma-v0')
obs, info = env.reset(seed=42)

print(f'Agents: {env.n_agents}')
print(f'Observation shape: {obs.shape}')
print(f'Action space: {env.action_space}')
```

```python
total_reward = np.zeros(env.n_agents)

for step in range(50):
    action = env.action_space.sample()
    obs, reward, terminated, truncated, info = env.step(action)
    total_reward += reward
    if terminated or truncated:
        break

print(f'Steps taken: {step + 1}')
print(f'Cumulative reward per agent: {total_reward}')
env.close()
```

## Multi-agent (PettingZoo Parallel) API

The PettingZoo Parallel adapter exposes the same environments through a dictionary-keyed multi-agent interface.

```python
env = coopetition_gym.make_parallel('PlatformEcosystem-v0')
obs_dict, info = env.reset(seed=42)

print(f'Agents: {list(env.agents)}')
print(f'Observation per agent: {obs_dict[env.agents[0]].shape}')

actions = {a: env.action_space(a).sample() for a in env.agents}
obs_dict, rewards, terminations, truncations, info = env.step(actions)
print(f'Rewards: {rewards}')
env.close()
```

## Reward-type ablation

Each environment supports three reward configurations that vary the structural mutuality between agents while holding the underlying environment dynamics fixed.

```python
for mode in ['private', 'integrated', 'cooperative']:
    env = coopetition_gym.make('TrustDilemma-v0', reward_mode=mode)
    obs, info = env.reset(seed=42)
    obs, reward, _, _, _ = env.step(env.action_space.sample())
    print(f'reward_mode={mode:12s} -> reward={reward}')
    env.close()
```
